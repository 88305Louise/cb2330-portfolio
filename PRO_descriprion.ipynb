**Project overview**

This project aims to investigate wether it's possible to model the data presented in table 2 in (Llopart et al., 2025) according to a poisson distrubution and determine the accuracy of that model.

The program rendered values that where similar to the ones observed directly from the paper, which means that a possion distribution is a possible model for the data investigated. However, since only a few chromosome arms are presented in the data, the model might not represent the reality of other chromosome arms.
